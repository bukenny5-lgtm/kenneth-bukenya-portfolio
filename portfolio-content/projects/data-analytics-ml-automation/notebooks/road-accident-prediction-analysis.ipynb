```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.decomposition import PCA
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import VotingClassifier
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
```

```python
df_ml = df_clean.copy()
```

```python
df_ml["Hour"] = pd.to_datetime(
    df_ml["Time of Day"],
    format="%H:%M",
    errors="coerce"
).dt.hour
```

```python
df_ml.drop(
    columns=["Time of Day"],
    inplace=True
)
```

```python
df_ml[["Hour"]].head()
```

```python
y = df_ml["Accident Severity"]
```

```python
X = df_ml.drop(
    columns=[
        "Accident Severity",
        "Number of Fatalities",
        "Number of Casualties"
    ]
)
```

```python
print("Predictor shape:", X.shape)
print("Target shape:", y.shape)
```

```python
numerical_features = X.select_dtypes(
    include="number"
).columns.tolist()

categorical_features = X.select_dtypes(
    exclude="number"
).columns.tolist()
```

```python
print("Numerical features:")
print(numerical_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training predictors:", X_train.shape)
print("Testing predictors:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
numerical_transformer = Pipeline(
    steps=[
        (
            "missing_values",
            SimpleImputer(strategy="median")
        ),
        (
            "scaling",
            StandardScaler()
        )
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        (
            "missing_values",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoding",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_transformer,
            numerical_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ]
)
```

```python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Support Vector Machine": SVC(
        kernel="rbf",
        probability=True,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    )
}
```

```python
results = []
trained_models = {}

for model_name, model in models.items():

    model_pipeline = Pipeline(
        steps=[
            ("preprocessing", preprocessor),
            ("model", model)
        ]
    )

    # Train the model
    model_pipeline.fit(X_train, y_train)

    # Make predictions
    predictions = model_pipeline.predict(X_test)

    # Calculate evaluation scores
    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })

    trained_models[model_name] = model_pipeline
```

```python
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

results_df.round(4)
```

```python
plot_results = results_df.copy()

plot_results["Accuracy"] = (
    plot_results["Accuracy"] * 100
)

plot_results["F1 Score"] = (
    plot_results["F1 Score"] * 100
)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=plot_results,
    x="Accuracy",
    y="Model",
    hue="Model",
    palette="viridis",
    legend=False
)

plt.title("Classification Model Accuracy")
plt.xlabel("Accuracy (%)")
plt.ylabel("Model")
plt.xlim(0, 100)
plt.show()
```

```python
best_model_name = results_df.loc[0, "Model"]

best_model = trained_models[best_model_name]

print("Best model:", best_model_name)
```

```python
best_predictions = best_model.predict(X_test)
```

```python
print(
    classification_report(
        y_test,
        best_predictions,
        zero_division=0
    )
)
```

```python
severity_labels = [
    "Minor",
    "Serious",
    "Fatal"
]

confusion = confusion_matrix(
    y_test,
    best_predictions,
    labels=severity_labels
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    confusion,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=severity_labels,
    yticklabels=severity_labels
)

plt.title(
    f"Confusion Matrix: {best_model_name}"
)

plt.xlabel("Predicted Severity")
plt.ylabel("Actual Severity")
plt.show()
```

```python
pca_pipeline = Pipeline(
    steps=[
        ("preprocessing", preprocessor),

        (
            "pca",
            PCA(
                n_components=0.95,
                random_state=42
            )
        ),

        (
            "model",
            LogisticRegression(
                max_iter=2000,
                random_state=42
            )
        )
    ]
)
```

```python
pca_pipeline.fit(
    X_train,
    y_train
)
```

```python
pca_predictions = pca_pipeline.predict(
    X_test
)
```

```python
pca_accuracy = accuracy_score(
    y_test,
    pca_predictions
)

pca_f1 = f1_score(
    y_test,
    pca_predictions,
    average="weighted"
)

print("PCA accuracy:", round(pca_accuracy, 4))
print("PCA F1 score:", round(pca_f1, 4))
```

```python
number_of_components = (
    pca_pipeline.named_steps["pca"].n_components_
)

print(
    "Number of PCA components:",
    number_of_components
)
```

```python
voting_classifier = VotingClassifier(
    estimators=[
        (
            "logistic",
            LogisticRegression(
                max_iter=2000,
                random_state=42
            )
        ),
        (
            "svm",
            SVC(
                probability=True,
                random_state=42
            )
        ),
        (
            "random_forest",
            RandomForestClassifier(
                n_estimators=300,
                random_state=42,
                n_jobs=-1
            )
        )
    ],
    voting="soft"
)
```

```python
ensemble_pipeline = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        ("model", voting_classifier)
    ]
)
```

```python
ensemble_pipeline.fit(
    X_train,
    y_train
)

ensemble_predictions = ensemble_pipeline.predict(
    X_test
)
```

```python
ensemble_accuracy = accuracy_score(
    y_test,
    ensemble_predictions
)

ensemble_f1 = f1_score(
    y_test,
    ensemble_predictions,
    average="weighted"
)

print(
    "Ensemble accuracy:",
    round(ensemble_accuracy, 4)
)

print(
    "Ensemble F1 score:",
    round(ensemble_f1, 4)
)
```
